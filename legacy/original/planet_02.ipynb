<h3>artools: pxscan()
</h3>
xscan di una sorgente del sistema solare

```python
from artools import pxscan, savetrack
import matplotlib.pyplot as plt
```

```python
#'mars', 'jupiter', 'saturn', 'venus', 'sun', 'moon'
obj='saturn'

# dt [s] è l'intervallo di tempo tra un punto della scansione e il successivo
dt=.5

# N è il numero di punti per ogni riga
# deve essere dispari, altrimenti gli viene aggiunto 1
N=20

# ANG è la semi-larghezza della mappa in gradi
ANG=0.5

epoch = '2025-11-19 22:12:00'
#epoch = 'NOW'

t,mjd,az,el = pxscan(obj,epoch,dt,N,ANG)
print('vel media:',round((ANG*2)/(dt*N),3),'deg/s')

savetrack('test.txt',t,az,el)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(az, el, '*-')  
plt.xlabel('Az (deg)')  
plt.ylabel('El (deg)')  
plt.grid()
```
