<h3>artools: ptrack()</h3>
tracking di un pianeta

```python
import sys
import matplotlib.pyplot as plt
from artools import ptrack, savetrack
```

```python
#'mars', 'jupiter', 'saturn', 'venus', 'sun', 'moon'
obj='venus'

# dt [s] è l'intervallo di tempo tra un punto della scansione e il successivo
dt=60

# N è il numero di punti per ogni riga
# deve essere dispari, altrimenti gli viene aggiunto 1
N=1000

epoch = '2026-08-06 18:30:00'
#epoch = 'NOW'

t,mjd,az,el=ptrack(obj,epoch,dt,N)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(mjd, az, '*')  
plt.xlabel('mjd')  
plt.ylabel('Az (deg)')  
plt.grid() 
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(mjd, el, '*')  
plt.xlabel('mjd')  
plt.ylabel('El (deg)')  
plt.grid() 
```

savetrack salva i dati nel formato ACU, mentre savetrack2 in un formato comodo per Matlab

```python
savetrack('test2.txt',t,az,el)
```
