<h3>artools: sxscan(tle_string,epoch,dt,N,ANG,REF)</h3>
satellite xscan


```python
import matplotlib.pyplot as plt
from artools import download_tle, tle ,sxscan, savetrack
```

```python
SAT="ASTRA 1L"
epoch = '2023-08-31 14:17:00'
dt=1
N=35
ANG=.15
REF=False
```

```python
download_tle()
tle_string=tle(SAT)
```

```python
t,mjd,az,el=sxscan(tle_string,epoch,dt,N,ANG,REF)
print(t[0], 'UTC')
savetrack('test.txt',t,az,el)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(az, el, '-*')  
plt.xlabel('Az (deg)')  
plt.ylabel('El (deg)')  
plt.grid() 
```
