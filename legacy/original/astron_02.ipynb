<h3>artools: xscan(source,epoch,dt,N,ANG,PRESSURE,TEMPERATURE,RELATIVE_HUMIDITY)
</h3>
xscan di una sorgente radio astronomica

```python
from artools import xscan, savetrack
import matplotlib.pyplot as plt
```

```python
source='W3(OH)'
epoch = '2023-08-08 08:00:00'
#epoch = 'NOW'

dt=1
N=10
ANG=1

#t,mjd,az,el = xscan(source,epoch,dt,N,ANG)
t,mjd,az,el = xscan(source,epoch,dt,N,ANG,PRESSURE=920,TEMPERATURE=20,RELATIVE_HUMIDITY=65)

savetrack('test.txt',t,az,el)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(az, el, '*-')  
plt.xlabel('Az (deg)')  
plt.ylabel('El (deg)')  
plt.grid()
```
