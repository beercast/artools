<h3>artools: track(source,epoch,dt,N,PRESSURE,TEMPERATURE,RELATIVE_HUMIDITY)</h3>
tracking di una sorgente radio astronomica

```python
import sys
import matplotlib.pyplot as plt
from artools import track, savetrack
```

```python
source='W3(OH)'
epoch = 'NOW'
#epoch = '2025-12-01 06:00:00'
dt=600
N=120
PRESSURE=980
TEMPERATURE=26
RELATIVE_HUMIDITY = 56
```

```python
t,mjd,az,el=track(source,epoch,dt,N,PRESSURE,TEMPERATURE,RELATIVE_HUMIDITY)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(mjd, az, '*')  
plt.xlabel('mjd')  
plt.ylabel('Az (deg)')  
plt.grid() 
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(mjd, el, '*')  
plt.xlabel('mjd')  
plt.ylabel('El (deg)')  
plt.grid() 
```

savetrack salva i dati nel formato ACU, mentre savetrack2 in un formato comodo per Matlab

```python
savetrack('test2.txt',t,az,el)
```
