<h3>artools: pmap(source,epoch,dt,N,ANG,PRESSURE,TEMPERATURE,RELATIVE_HUMIDITY)</h3>

crea una mappa da oggetto del sistema solare

```python
import matplotlib.pyplot as plt
import numpy as np
from artools import pmap, savetrack
```

```python
#'mars', 'jupiter', 'saturn', 'venus', 'sun', 'moon'
obj='moon'

# dt [s] è l'intervallo di tempo tra un punto della scansione e il successivo
dt=.25

# N è il numero di punti per ogni riga
# deve essere dispari, altrimenti gli viene aggiunto 1
N=25

# ANG è la semi-larghezza della mappa in gradi
# la mappa è "quadrata" cioè larghezza=altezza
# 1 deg = circa 5 beams
ANG=0.5

epoch = '2026-08-31 22:30:00'
#epoch = 'NOW'
```

```python
t,mjd,az,el=pmap(obj,epoch,dt,N,ANG)

N0=int((N+1)*np.floor(N/2))
print('ns:',N0+1,'azs:',round(az[N0],3),'els:',round(el[N0],3))
print('vel media:',round((ANG*2)/(dt*N),3),'deg/s')
print('durata scansione:',round((mjd[N*N-1]-mjd[0])*24*60,3),'min')

#salva nel formato della ACU
savetrack('map.txt',t,az,el)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(az, el, '-*')  
plt.xlabel('Az (deg)')  
plt.ylabel('El (deg)')  
plt.grid() 
```
