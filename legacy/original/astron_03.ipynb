<h3>artools: amap(source,epoch,dt,N,ANG,PRESSURE,TEMPERATURE,RELATIVE_HUMIDITY)</h3>    
mappa di una sorgente radio astronomica

```python
import matplotlib.pyplot as plt
import numpy as np
from artools import amap, savetrack
```

```python
source='W3(OH)'
#epoch = '2023-08-08 08:00:00'
epoch = 'NOW'
dt=1
N=25
ANG=0.4
```

```python
t,mjd,az,el = amap(source,epoch,dt,N,ANG,PRESSURE=920,TEMPERATURE=20,RELATIVE_HUMIDITY=65)

N0=int((N+1)*np.floor(N/2))
print('n   az_sorg            el_sorg')
print(N0+1,az[N0],el[N0])

savetrack('test.txt',t,az,el)
```

```python
plt.close()  
fig = plt.figure(figsize=(8,8))  
plt.plot(az, el, '-*')  
plt.xlabel('Az (deg)')  
plt.ylabel('El (deg)')  
plt.grid() 
```
