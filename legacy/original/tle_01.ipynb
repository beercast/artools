<h3>artools: strack(tle_string,epoch,dt,N,REF)</h3>

tracking di un satellite

```python
import matplotlib.pyplot as plt
from artools import download_tle, tle ,strack, savetrack2, savetrack
```

In questo esempio la stringa della tle (tle_string) è inserita "a mano".
Più avanti verrà mostrato come inserirla automaticamente e come aggiornarla

```python
epoch = '2023-08-31 12:11:00'
dt=600
N=144
tle_string='''EUTELSAT HOTBIRD 13B
1 29270U 06032A   23243.11624506  .00000071  00000+0  00000+0 0  9996
2 29270   0.0872  56.3198 0003129 117.8334 219.6987  1.00271788 62625'''
```

```python
t,mjd,az,el,dist=strack(tle_string,epoch,dt,N,True)
print(t[0], 'UTC')
print('distance {:.0f} km'.format(dist)) 
```

```python
fig = plt.figure(figsize=(8,8))  
plt.plot(az, el, '*')  
plt.xlabel('Az (deg)')  
plt.ylabel('El (deg)')  
plt.grid() 
```

Non è necessario inserire le tle del satellite "a mano" perché sono già contenute nel file norad_tle.txt

download_tle() fa il refresh del file norad_tle.txt

tle(satellite) carica dal file norad_tle.txt le tle del satellite

```python
download_tle()
tle_string=tle('ASTRA 1L')

dt=1
N=3*900

t,mjd,az,el,dist=strack(tle_string,'NOW',dt,N)

print(t[0], 'UTC')
savetrack2('test.txt',mjd,az,el)
```
